# pre processing

```python
#202431002_david Lie Agatha
df = pd.read_csv('iris-flower.csv', sep=",")
df
```

```python
#202431002_david Lie Agatha
df.info()
```

```python
#202431002_david Lie Agatha
df['Id'].unique()
```

```python
#202431002_david Lie Agatha
df['Species'].unique()
```

```python
#202431002_david Lie Agatha
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['Species'] = le.fit_transform(df['Species'])
print(df['Species'].unique())

for i, label in enumerate(le.classes_):
    print(f"{i}={label}")
```

```python
#202431002_david Lie Agatha
df
```

# modelling

```python
#202431002_david Lie Agatha
x= df.drop(['Id', 'Species'], axis=1)
x
```

```python
#202431002_david Lie Agatha
y = df['Species']
y
```

```python
#202431002_david Lie Agatha
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.25, random_state=0)
```

```python
#202431002_david Lie Agatha
from sklearn.tree import DecisionTreeClassifier

classifier = DecisionTreeClassifier()
classifier.fit(x_train, y_train)
```

```python
#202431002_david Lie Agatha
from sklearn import tree

plt.subplots(figsize = (15,10))
tree.plot_tree(classifier, fontsize = 10)
plt.show()
```

```python
#202431002_david Lie Agatha
from sklearn.tree import DecisionTreeClassifier

classifier = DecisionTreeClassifier(criterion='entropy', max_depth=5, random_state=42)
classifier.fit(x_train, y_train)
```

```python
#202431002_david Lie Agatha
from sklearn import tree

plt.subplots(figsize = (15,10))
tree.plot_tree(classifier, fontsize = 10)
plt.show()
```

```python
#202431002_david Lie Agatha
depth = classifier.get_depth()
depth
```

```python
#202431002_david Lie Agatha
from sklearn.metrics import accuracy_score

y_pred = classifier.predict(x_test)
akurasi = accuracy_score(y_test, y_pred)
print(akurasi)
```

```python
#202431002_david Lie Agatha
y_pred = classifier.predict([[3.3, 5,7,0.7]])
print(y_pred)
```
